### Exploratory Data Analysis

#### 1. Movies Vs TV Shows

```python
df_clean["type"].value_counts()
```

```python
type_counts = df_clean["type"].value_counts()

sns.barplot(
    x=type_counts.index,
    y=type_counts.values
)

plt.title("Number of Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Number of Titles")
plt.show()
```

#### 2. Which year did Netflix add the most titles

```python
df_clean["date_added"].dt.year.value_counts().sort_index()
```

```python
# Number of titles added to Netflix each year

titles_added_year = df_clean["date_added"].dt.year.value_counts().sort_index()

plt.figure(figsize=(12, 6))

plt.plot(titles_added_year.index, titles_added_year.values, marker="o")

plt.title("Number of Netflix Titles Added by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.show()
```

#### 3. Which countries has the most Netflix Titles.

```python
country_split = df_clean["country"].str.split(", ")
```

```python
country_exploded = country_split.explode()
```

```python
country_counts = country_exploded.value_counts()

country_counts.head(10)
```

```python
top_countries = country_counts.drop("Unknown").head(10)

top_countries
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title("Top 10 Countries by Number of Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

#### 4. Most Common Genres.

```python
df_clean["listed_in"].head(10)
```

```python
genre_split = df_clean["listed_in"].str.split(", ")
```

```python
genre_exploded = genre_split.explode()
```

```python
genre_counts = genre_exploded.value_counts()
genre_counts.head(10)
```

```python
# Visualize the top 10 Genres
top_genres = genre_counts.head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title("Top 10 Genres on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```

#### 5. Movies Vs TV Shows over the years

```python
content_by_year = (
    df_clean
    .dropna(subset=["date_added"])
    .assign(year=df_clean["date_added"].dt.year)
    .groupby(["year", "type"])
    .size()
    .unstack(fill_value=0)
)

content_by_year
```

```python
# Visualize
content_by_year.plot(
    kind="line",
    marker="o",
    figsize=(12, 6)
)

plt.title("Movies and TV Shows Added to Netflix by Year")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.legend(title="Type")

plt.show()
```

##### Lets create the overall distribution of Movies and TV shows available in Netflix.

```python
# Count Movies and TV Shows
type_counts = df_clean["type"].value_counts()

type_counts
```

```python
type_percentage = (type_counts / type_counts.sum()) * 100

type_percentage
```

```python
# visualize
plt.figure(figsize=(7, 7))

plt.pie(
    type_counts,
    labels=type_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Movies and TV Shows on Netflix")

plt.show()
```

##### Let's identify the countries with the highest number of titles in the dataset.

```python
country_counts = (
    df_clean[df_clean["country"] != "Unknown"]["country"]
    .value_counts()
    .head(10)
)

country_counts
```

```python
# Visualize
plt.figure(figsize=(10, 6))

country_counts.head().sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

#### 6. Most common ratings on Netflix.

```python
# Count the Content Ratings
rating_counts = df_clean["rating"].value_counts()

rating_counts
```

```python
# visualize the top 10 ratings
top_ratings = rating_counts.head(10)

plt.figure(figsize=(10, 6))

top_ratings.sort_values().plot(kind="barh")

plt.title("Top 10 Content Ratings on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")

plt.show()
```

```python
# calculating the percentage.
rating_percentage = (rating_counts / rating_counts.sum()) * 100

rating_percentage.head(10)
```
